# Matpltolib Properties

```python
import matplotlib.pyplot as plt

x =[1,2,3,4,5]
y = [10,20,30,35,40]

plt.plot(x,y)
plt.show()
```

Customized chart

```python
plt.plot(x, y, color='green',linestyle = 'dashed',linewidth = 2)
plt.show()
```

Advanced Line Chart

```python


x =[1,2,3,4,5]
y1 = [10,20,30,35,40]
y2 = [20,25,35,45,50]

plt.plot(x, y1, label='Sales 2024')
plt.plot(x, y2, label='Sales 2025')

plt.title("Yearly Sales")
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

Bar chart

```python
x =["Islamabad","Quetta","Peshawar","Lahore","Karachi"]
y1 = [10,20,30,35,45]


plt.bar(x,y1)
plt.title("City Population")
plt.xlabel('Cities')
plt.ylabel('Population')

plt.show()
```

Pie chart

```python
brands = ["iphone","samsung","vivo","google","huaweii"]
sales_per_year = [100,80,150,200,220]

plt.pie(sales_per_year, labels=brands,autopct='%1.1f%%')

plt.title("Brands Sales per Year")
plt.show()
```

Scatter plot

```python
x =[1,2,3,4,5]
y = [10,20,30,35,40]
plt.scatter(x,y)
plt.title('Scatter plot Example')
plt.show()
```

```python
categories =['mon','tues','wed','thurs','fri','sat','sun']
sales = [10,20,30,35,40,45,46]
 # first plot bar chart
plt.subplot(1,1,1)
plt.bar(categories, sales)
plt.title("Daily Sales ")

plt.xlabel("Week Days")
plt.ylabel(sales)

plt.show()


#second plot scatter chart

plt.subplot(1,2,2)
plt.scatter(categories,sales)
plt.title("Daily Sales")

plt.xlabel("Week Days")
plt.ylabel(sales)

plt.show()
```
